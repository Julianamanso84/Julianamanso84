```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv('/content/ecommerce_preparados.csv')
print(df.head().to_string())

#HISTOGRAMA 1- QUANTIDADE_VENDIDOS
plt.hist(df['Qtd_Vendidos'])
plt.show()

#Parâmetros Histograma
plt.figure(figsize=(10, 5))
plt.hist(df['Qtd_Vendidos'], bins=15, color='skyblue', alpha= 0.9)
plt.title('Histograma de Quantidade Vendida')
plt.xlabel('Quantidade Vendida')
plt.ylabel('Frequência')
plt.grid(True)
plt.show()

# Dispersão
plt.figure(figsize=(5,5))
plt.scatter(df['Qtd_Vendidos_Cod'],df['Preço'], color = 'red', alpha = 0.9)
plt.title('Dispersão de Quantidade Vendida x Preço')
plt.xlabel('Qtd_Vendidos_Cod')
plt.ylabel('Preço')
plt.show()

# Dispersão com Seaborn
plt.figure(figsize=(5,5))
sns.scatterplot(x='Qtd_Vendidos_Cod', y='Preço',data=df, hue='Qtd_Vendidos_Cod', palette='viridis')
plt.title('Dispersão de Quantidade Vendida x Preço')
plt.xlabel('Qtd_Vendidos_Cod')
plt.ylabel('Preço')
plt.show()

# Mapa de calor
corr=df[['Qtd_Vendidos_Cod','Preço']].corr()
plt.figure(figsize=(5,5))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Mapa de Calor Qtdade Vendida_Preço')
plt.show()

#Barras
plt.figure(figsize=(10, 5))
df['Qtd_Vendidos_Cod'].value_counts().plot(kind='bar', color='skyblue')
plt.title('Quantidade de Vendas por Preço')
plt.xlabel('Qtd_Vendidos_Cod')
plt.ylabel('Preço')
plt.show()

#Pizza
plt.figure(figsize=(10, 5))
counts = df['Qtd_Vendidos_Cod'].value_counts()
x = counts.index
y = counts.values
plt.pie(y, labels=x,autopct='%.1f%%', startangle=90)
plt.title('Distribuição de Qtd_Vendidos_Cod')
plt.show()
```
